## imports + saved data 
- reload preprocessed data from prev notebook
- load 3 diff models (good for comparison)

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, f1_score, recall_score, RocCurveDisplay
from xgboost import XGBClassifier
import numpy as np
import pandas as pd
import joblib

X_train = np.load('../data/processed/X_train_resampled.npy')
X_test  = np.load('../data/processed/X_test_processed.npy')
y_train = np.load('../data/processed/y_train_resampled.npy')
y_test  = np.load('../data/processed/y_test.npy')

preprocessor = joblib.load('../models/preprocessor.pkl')

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
```

## logistic regression omdel

```python
lr_clf = LogisticRegression(random_state=42, max_iter=1000)
lr_clf.fit(X_train, y_train)
# y_pred = lr_clf.predict(X_test)
# y_proba = lr_clf.predict_proba(X_test)
# accuracy = lr_clf.score(X_test, y_test)
```

## random forest

```python
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)
```

## xgbclassifier

```python
xgb_clf = XGBClassifier(random_state=42, eval_metric='logloss')
xgb_clf.fit(X_train, y_train)

# y_pred = xgb_clf.predict(X_test)
# print("Accuracy:", accuracy_score(y_test, y_pred))
```

## accuracy eval
- roc and f1

```python
#logistic regression (lr_clf, rf_clf, xbg_clf)
y_pred  = lr_clf.predict(X_test)
y_proba = lr_clf.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))

#random forest
y_pred  = rf_clf.predict(X_test)
y_proba = rf_clf.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))

#xgboost
y_pred  = xgb_clf.predict(X_test)
y_proba = xgb_clf.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
```

# plotted ROC curves 
- compare models

```python
fig, ax = plt.subplots(figsize=(8, 6))

RocCurveDisplay.from_estimator(lr_clf, X_test, y_test, ax = ax, name='Logistic Regression')
RocCurveDisplay.from_estimator(rf_clf, X_test, y_test, ax=ax, name='Random Forest')
RocCurveDisplay.from_estimator(xgb_clf, X_test, y_test, ax=ax, name='XGBoost')

plt.title('ROC Curves for Logistic Regression, Random Forest, and XGBoost')
plt.grid(alpha=0.3)

plt.show()
```

```python
#comparison table

models = {
    'Logistic Regression': lr_clf,
    'Random Forest':       rf_clf,
    'XGBoost':             xgb_clf,
}

rows = []
for name, model in models.items():
    y_pred  = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:, 1]
    rows.append({
        'model':     name,
        'accuracy':  accuracy_score(y_test, y_pred),
        'f1 (churn)': f1_score(y_test, y_pred),
        'roc-auc':   roc_auc_score(y_test, y_proba),
        'recall (churn)': recall_score(y_test, y_pred),
    })

results_df = pd.DataFrame(rows).set_index('model').round(3)
print(results_df)
```

```python
# saving models to use for later 

import os
os.makedirs('../models', exist_ok=True)

joblib.dump(lr_clf,  '../models/lr_model.pkl')
joblib.dump(rf_clf,  '../models/rf_model.pkl')
joblib.dump(xgb_clf, '../models/xgb_model.pkl')

print("all 3 models saved to models/")
```
